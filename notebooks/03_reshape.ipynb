# Step 03.3 — Reshape

这一小步只回答一个问题：**为什么同一批数字可以用不同的 shape 表示？**

````python
import mlx.core as mx

x = mx.array([
    [1, 2, 3],
    [4, 5, 6],
])

print(x)
print("shape:", x.shape)
````

`x.shape = (2, 3)`，所以总元素数是 `2 × 3 = 6`。

`reshape` 可以重新组织这 6 个位置，但不能凭空增加或删除元素。

````python
y = x.reshape(3, 2)

print(y)
print("shape:", y.shape)
````

结果会把原来的 6 个元素按原有顺序重新排成 `3 × 2`：

```text
[[1, 2, 3],      [[1, 2],
 [4, 5, 6]]  →    [3, 4],
                  [5, 6]]
```

注意：`reshape` 不是矩阵转置。

````python
z = x.reshape(6)

print(z)
print("shape:", z.shape)
````

这里 `(2, 3) → (6,)` 仍然合法，因为总元素数始终是 6。

因此 reshape 的最基本规则是：

`旧 shape 各维长度乘积 = 新 shape 各维长度乘积`。

以后 Multi-Head Attention 里的 `(B, T, C) → (B, T, H, D)` 就会用到同一个规则，其中 `C = H × D`。
