# Step 03.4 — Reshape vs Transpose

这一小步只区分两个操作：

- `reshape`：重新解释 shape
- `transpose`：交换 axis 的顺序

````python
import mlx.core as mx

x = mx.array([
    [1, 2, 3],
    [4, 5, 6],
])

print(x)
print("shape:", x.shape)
````

原始 `x` 是 2 行 3 列：`shape = (2, 3)`。

````python
a = x.reshape(3, 2)

print(a)
print("reshape shape:", a.shape)
````

`reshape(3, 2)` 只是按原来的元素顺序重新分组：

```text
[[1, 2, 3],      [[1, 2],
 [4, 5, 6]]  ->   [3, 4],
                  [5, 6]]
```

````python
b = x.transpose(1, 0)

print(b)
print("transpose shape:", b.shape)
````

`transpose(1, 0)` 是把 axis 0 和 axis 1 对调。

```text
[[1, 2, 3],      [[1, 4],
 [4, 5, 6]]  ->   [2, 5],
                  [3, 6]]
```

所以虽然两者最后 shape 都是 `(3, 2)`，里面的数字排列完全不同。

### 本步只记住

- `reshape`：元素顺序保持，重新切块
- `transpose`：axis 顺序改变

以后 Multi-Head Attention 会连续用到两者：

`(B, T, C) -> reshape -> (B, T, H, D) -> transpose -> (B, H, T, D)`
