## Step 02.5 — MLX / Metal probe

这一小步只确认三件事：

1. MLX 能导入；
2. Metal 后端可用；
3. MLX 默认计算设备是 GPU。

```python
from importlib.metadata import version
import mlx.core as mx

print("MLX version    :", version("mlx"))
print("Metal available:", mx.metal.is_available())
print("Default device :", mx.default_device())
print("Device info    :", mx.device_info())
```

如果环境正确，`Metal available` 应为 `True`，默认设备应显示 GPU。

```python
x = mx.array([[1.0, 2.0], [3.0, 4.0]])
w = mx.array([[2.0, 0.0], [0.0, -1.0]])

y = x @ w
mx.eval(y)

print(y)
```

这里第一次出现 `mx.eval(y)`。现在只把它理解成：**要求 MLX 真正把 `y` 算出来。**

为什么 MLX 不一定在写下 `y = x @ w` 的瞬间就计算，会在后面专门讲 Lazy Evaluation。
