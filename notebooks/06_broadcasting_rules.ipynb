# Step 03.6 — Broadcasting Rules

正式规则很简单：**从最后一个维度开始，逐维向左比较。**

每一对维度必须满足下面至少一个条件：

1. 两个数字相同；
2. 其中一个数字是 `1`；
3. 某一边缺少这个维度。

### 例 1：`(2, 3) + (3,)`

从右往左比较：

```text
(2, 3)
    (3)
     ↑
3 和 3 相同 -> 可以
```

较短的 `(3,)` 左边缺少一维，因此可以把它理解成 `(1, 3)`，再沿最外层广播到 `(2, 3)`。

````python
import mlx.core as mx

a = mx.ones((2, 3))
b = mx.array([10, 20, 30])

print("a shape:", a.shape)
print("b shape:", b.shape)
print("result shape:", (a + b).shape)
````

### 例 2：`(2, 3) + (2, 1)`

从右往左比较：

```text
(2, 3)
(2, 1)
 ↑  ↑
 2=2
 3 和 1 -> 因为有一个是 1，所以可以
```

`1` 可以扩展成这一轴需要的长度。

````python
c = mx.ones((2, 3))
d = mx.array([[10], [20]])

print(c + d)
print("result shape:", (c + d).shape)
````

### 例 3：`(2, 3) + (2,)` 为什么不行？

从最后一维开始：

```text
(2, 3)
    (2)
     ↑
3 和 2 不相同，而且谁都不是 1
```

所以广播失败。

````python
bad = mx.array([10, 20])

try:
    print(a + bad)
except Exception as e:
    print(type(e).__name__ + ":", e)
````
