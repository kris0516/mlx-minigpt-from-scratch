# Step 03.5 — Broadcasting

Broadcasting 的核心：**小 tensor 可以在缺少的维度上被“逻辑复制”，从而和大 tensor 做逐元素运算。**

````python
import mlx.core as mx

x = mx.array([
    [1, 2, 3],
    [4, 5, 6],
])

b = mx.array([10, 20, 30])

print("x shape:", x.shape)
print("b shape:", b.shape)
````

`x.shape = (2, 3)`，`b.shape = (3,)`。

做 `x + b` 时，`b` 会被理解成对两行都使用同一组 `[10, 20, 30]`。

````python
y = x + b

print(y)
print("shape:", y.shape)
````

逻辑上相当于：

```text
[[1, 2, 3],      [[10, 20, 30],
 [4, 5, 6]]  +    [10, 20, 30]]
```

所以结果：

```text
[[11, 22, 33],
 [14, 25, 36]]
```

### 本步只记住

Broadcasting 不是把数据真的手工复制很多份，而是让框架按兼容规则把较小 tensor 当作可以扩展来参与逐元素运算。
